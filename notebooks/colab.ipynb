## Configure

A *quarter five-spot*: water injected in one corner of a square,
oil produced in the other. Rates are **signed** -- positive injects,
negative produces -- and everything else
(permeability, porosity, fluid properties) is left at its default.


```python
import matplotlib.pyplot as plt
import numpy as np

from minires import ResSim

model = ResSim(Lx=1, Ly=1, Nx=32, Ny=32, wells=[
    dict(name="Inj",  xy=[0, 0], rate=+1),
    dict(name="Prod", xy=[1, 1], rate=-1),
])
model
```

## Simulate

`sim` returns the saturation and pressure trajectories,
`(nSteps + 1, Nxy)` each.


```python
nSteps = 28
S, P = model.sim(0.7/nSteps, nSteps, S0=np.zeros(model.Nxy))
```

## Plot the saturation field


```python
fig, ax = plt.subplots(figsize=(5, 4.2))
model.plt_field(ax, S[-1], "oil");
```

## Animate it


```python
from IPython.display import HTML

prd = model.xy2ind(*model.wells.xy[1])  # the producer's cell
HTML(model.anim(S, S[1:, [prd]]).to_jshtml())
```
